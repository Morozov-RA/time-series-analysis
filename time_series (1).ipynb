```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller

df = pd.read_csv('monthly-sales-of-company-x-jan-6.csv')

plt.figure(figsize=(12,4))
plt.plot(df['Count'])
plt.title('Monthly Sales of Company X - Original Series')
plt.grid()
plt.show()

result = adfuller(df['Count'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

```python
df['diff1'] = df['Count'].diff()

plt.figure(figsize=(12,4))
plt.plot(df['diff1'])
plt.title('Monthly Sales of Company X - First Difference')
plt.grid()
plt.show()

result = adfuller(df['diff1'].dropna())

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

```python
df['diff2'] = df['diff1'].diff()

plt.figure(figsize=(12,4))
plt.plot(df['diff2'])
plt.title('Monthly Sales of Company X - Second Difference')
plt.grid()
plt.show()

result = adfuller(df['diff2'].dropna())

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

### Вывод

Исходный временной ряд является нестационарным (p-value = 0.9889).

После применения первой разности ряд остаётся нестационарным (p-value = 0.0955).

После применения второй разности значение p-value стало равно 2.09e-15, что меньше уровня значимости 0.05.

Следовательно, итоговый стационарный ряд получен после второго дифференцирования.

```python
df = pd.read_csv('monthly-boston-armed-robberies-j.csv')

# Исходный ряд
plt.figure(figsize=(12,4))
plt.plot(df['Count'])
plt.title('Boston Armed Robberies - Original Series')
plt.grid()
plt.show()

# ADF-тест
result = adfuller(df['Count'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

```python
# Первая разность
df['diff1'] = df['Count'].diff()

plt.figure(figsize=(12,4))
plt.plot(df['diff1'])
plt.title('Boston Armed Robberies - First Difference')
plt.grid()
plt.show()

result = adfuller(df['diff1'].dropna())

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

### Вывод

Исходный временной ряд является нестационарным (p-value = 0.9943).

После применения первой разности значение p-value составило 6.45e-11, что меньше уровня значимости 0.05.

Следовательно, итоговый стационарный ряд получен после первого дифференцирования.

```python
df = pd.read_csv('international-airline-passengers.csv')

plt.figure(figsize=(12,4))
plt.plot(df['Count'])
plt.title('International Airline Passengers - Original Series')
plt.grid()
plt.show()

result = adfuller(df['Count'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

```python
import numpy as np

# Логарифмирование
df['log'] = np.log(df['Count'])

# Сезонная разность
df['seasonal_diff'] = df['log'].diff(12)

# Обычная разность
df['final_series'] = df['seasonal_diff'].diff()

plt.figure(figsize=(12,4))
plt.plot(df['final_series'])
plt.title('International Airline Passengers - Final Stationary Series')
plt.grid()
plt.show()

result = adfuller(df['final_series'].dropna())

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

### Вывод

Исходный временной ряд является нестационарным (p-value = 0.9919) и содержит тренд и сезонность.

Для приведения ряда к стационарному виду были выполнены:
1. Логарифмирование.
2. Сезонное дифференцирование с периодом 12.
3. Обычное дифференцирование.

После преобразований значение p-value составило 0.0002486, что меньше 0.05.

Следовательно, итоговый стационарный ряд получен после логарифмирования, сезонного и обычного дифференцирования.

```python
df = pd.read_csv('mean-monthly-air-temperature-deg.csv')

plt.figure(figsize=(12,4))
plt.plot(df['Deg'])
plt.title('Mean Monthly Air Temperature - Original Series')
plt.grid()
plt.show()

result = adfuller(df['Deg'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])
```

### Вывод

Исходный временной ряд уже является стационарным.

Значение p-value теста Дики–Фуллера равно 0.0170, что меньше уровня значимости 0.05.

Дополнительные преобразования не требуются. В качестве итогового стационарного ряда используется исходный ряд.
